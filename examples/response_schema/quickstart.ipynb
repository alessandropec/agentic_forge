# Response Schema Quickstart

This notebook shows how declarative schema YAML becomes a runtime Pydantic model plus prompt-ready formatting instructions.

What you will do:

- forge a flat schema
- load the same schema again through `ref`
- forge a nested schema with objects and lists
- inspect the generated format instructions used by prompt-based workflows


```python
import sys
from pathlib import Path

NOTEBOOK_DIR = Path.cwd().resolve()
PROJECT_ROOT = next((p for p in [NOTEBOOK_DIR, *NOTEBOOK_DIR.parents] if (p / 'agentic_forge').exists()), NOTEBOOK_DIR)
if str(PROJECT_ROOT) not in sys.path:
    sys.path.insert(0, str(PROJECT_ROOT))

RESP_DIR = PROJECT_ROOT / 'examples' / 'response_schema'
MATERIALS_DIR = RESP_DIR / 'materials'

from agentic_forge import forge
from agentic_forge.structured_output import ForgedResponseSchema
```

## Step 1: Start with the simple schema

The first files demonstrate the smallest useful response schema shape:

- `response_schema_simple.yaml` defines the fields directly
- `response_schema_simple_ref.yaml` reuses that file through `ref`

This makes it easy to compare a normal schema material with a pure composition wrapper.


```python
simple_schema: ForgedResponseSchema = forge(MATERIALS_DIR / 'response_schema_simple.yaml')
simple_ref: ForgedResponseSchema = forge(MATERIALS_DIR / 'response_schema_simple_ref.yaml')

print(simple_schema.schema)
print(simple_schema.get_format_instruction())
print('Referenced schema name:', simple_ref.schema.__name__)
```

## Step 2: Inspect the nested schema

`response_schema_nested.yaml` shows how nested objects and list items are described declaratively. The generated format instructions are useful because they are the bridge between the schema definition and the prompt you send to a model.


```python
nested_schema: ForgedResponseSchema = forge(MATERIALS_DIR / 'response_schema_nested.yaml')
print(nested_schema.schema)
print(nested_schema.get_format_instruction(include_description=True)[:800])
```
