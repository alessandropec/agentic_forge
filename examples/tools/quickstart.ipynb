# Tools Quickstart

This notebook shows how declarative tool YAML becomes runtime callables with schemas, defaults, and injected resources.

What you will do:

- forge a single tool and inspect its OpenAI-style schema
- forge a toolset with shared resources
- call stateful and stateless tools
- inspect how defaults and argument definitions appear at runtime


```python
import sys
from pathlib import Path

NOTEBOOK_DIR = Path.cwd().resolve()
PROJECT_ROOT = next((p for p in [NOTEBOOK_DIR, *NOTEBOOK_DIR.parents] if (p / 'agentic_forge').exists()), NOTEBOOK_DIR)
if str(PROJECT_ROOT) not in sys.path:
    sys.path.insert(0, str(PROJECT_ROOT))

TOOLS_DIR = PROJECT_ROOT / 'examples' / 'tools'
MATERIALS_DIR = TOOLS_DIR / 'materials'

from agentic_forge import forge
from agentic_forge.tools import ForgedTool, ForgedToolset
```

## Step 1: Start with one reusable tool

`materials/single_tool.yaml` is intentionally small: it only points at `materials/tools/search.yaml`. This shows the separation between a minimal entrypoint material and a reusable tool definition.


```python
single_tool: ForgedTool = forge(MATERIALS_DIR / 'single_tool.yaml')
print(single_tool.to_openai_tool())
print(single_tool.invoke(query='agentic forge'))
```

## Step 2: Forge the toolset

`materials/toolset.yaml` adds the richer patterns used by agent workflows:

- a shared resource built once at forge time
- tools that inject runtime state
- tools that override defaults from a referenced YAML

The runtime output should make those wiring choices visible.


```python
toolset: ForgedToolset = forge(MATERIALS_DIR / 'toolset.yaml')
print(toolset.list_tools())
print(toolset.invoke('search_with_db', query='senate', limit=2))
print(toolset.invoke('remember_topic', topic='yaml', state={'visited_topics': []}))
print(toolset.invoke('answer_with_memory', question='How does it work?', state={'last_topic': 'yaml'}))
```
