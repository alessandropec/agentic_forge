# LLM Quickstart

This notebook shows how declarative provider YAML becomes runtime providers and selector-based managers.

What you will do:

- forge a multi-provider manager
- inspect the available `provider:model` selectors
- forge one provider directly
- optionally make a live model call if local credentials are configured


```python
import sys
from pathlib import Path

NOTEBOOK_DIR = Path.cwd().resolve()
PROJECT_ROOT = next((p for p in [NOTEBOOK_DIR, *NOTEBOOK_DIR.parents] if (p / 'agentic_forge').exists()), NOTEBOOK_DIR)
if str(PROJECT_ROOT) not in sys.path:
    sys.path.insert(0, str(PROJECT_ROOT))

import os

LLM_DIR = PROJECT_ROOT / 'examples' / 'llm'
MATERIALS_DIR = LLM_DIR / 'materials'

from agentic_forge import forge
from agentic_forge.llm import ForgedLlmManager, ForgedLlmProvider
```

## Step 1: Forge the manager

`materials/manager_multi_provider.yaml` aggregates the OpenAI, OpenRouter, and Ollama provider materials behind one runtime object. The printed selectors show the public names available to downstream components such as structured models.


```python
manager: ForgedLlmManager = forge(MATERIALS_DIR / 'manager_multi_provider.yaml')
for row in manager.list_models():
    print(row)

model = manager.get_model('default_openrouter:vision_chat')
print(type(model).__name__)
```

## Step 2: Forge a provider directly

Forging a single provider is useful when you want to inspect one model catalog in isolation or build models manually without going through the manager abstraction.


```python
provider: ForgedLlmProvider = forge(MATERIALS_DIR / 'provider_openrouter.yaml')
print(provider.models)
print(type(provider.get_model('vision_chat')).__name__)
```

## Step 3: Optional live call

Live calls are disabled by default so the notebook remains safe to run in a fresh clone. Enable them only after filling in `materials/api_keys.yaml` and confirming the selected provider is reachable from your environment.


```python
if os.environ.get('AGENTIC_FORGE_RUN_LIVE_LLM') == '1':
    live_model = manager.get_model('default_openrouter:vision_chat')
    print(live_model.invoke('Say hello in one sentence.'))
else:
    print('Live LLM call skipped. Set AGENTIC_FORGE_RUN_LIVE_LLM=1 after configuring your keys.')
```
